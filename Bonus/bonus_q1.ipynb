```python
from transformers import pipeline

# Initialize the pipeline
qa_pipeline = pipeline("question-answering")

# Define your context and question
context = """
The first mechanical computer, known as the Analytical Engine, was designed by Charles Babbage.
He is considered the father of the computer. His work laid the foundations for modern computing.
"""

question = "Who is known as the father of the computer?"

# Run the pipeline
result = qa_pipeline(question=question, context=context)

# Print the result
print("Answer:", result['answer'])
print("Confidence Score:", result['score'])
print("Start Index:", result['start'])
print("End Index:", result['end'])
```

```python
from transformers import pipeline

# Load a custom QA model
qa_pipeline = pipeline("question-answering", model="deepset/roberta-base-squad2")

# Context and question
context = """
The first mechanical computer, known as the Analytical Engine, was designed by Charles Babbage.
He is considered the father of the computer. His work laid the foundations for modern computing.
"""
question = "Who is known as the father of the computer?"

# Run the pipeline
result = qa_pipeline(question=question, context=context)

# Output results
print("Task 2 - Custom Model Output")
print("Answer:", result['answer'])
print("Score:", round(result['score'], 4))
print("Start Index:", result['start'])
print("End Index:", result['end'])
```

```python
# Custom context
custom_context = """
UCM is known for its strong Computer Science graduate programs.
Many international students enroll in UCM due to its affordable tuition and excellent faculty.
The Computer Science department is led by Dr. Smith.
"""

# Questions
questions = [
    "Why do international students choose UCM?",
    "Who leads the Computer Science department?"
]

# Run the pipeline for each question
print("\nTask 3 - Custom Context Questions")
for q in questions:
    result = qa_pipeline(question=q, context=custom_context)
    print(f"\nQuestion: {q}")
    print("Answer:", result['answer'])
    print("Score:", round(result['score'], 4))
    print("Start:", result['start'], "End:", result['end'])
```
