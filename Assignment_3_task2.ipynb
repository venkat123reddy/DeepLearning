```python
# ✅ Step 1: Install dependencies
!pip install -U spacy nltk

# ✅ Step 2: Download required models and data
import nltk
nltk.download('stopwords')

# Download spaCy English model
!python -m spacy download en_core_web_sm

# ✅ Step 3: Import libraries
import spacy
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer

# Load spaCy English model
nlp = spacy.load("en_core_web_sm")
stop_words = set(stopwords.words('english'))
stemmer = PorterStemmer()

# ✅ Step 4: Define preprocessing function
def nlp_preprocess(sentence):
    # Tokenization
    doc = nlp(sentence)
    tokens = [token.text for token in doc]
    print("Original Tokens:", tokens)

    # Remove stopwords and punctuation
    tokens_no_stop = [token.text for token in doc if token.text.lower() not in stop_words and token.is_alpha]
    print("Tokens Without Stopwords:", tokens_no_stop)

    # Apply stemming
    stemmed_words = [stemmer.stem(word) for word in tokens_no_stop]
    print("Stemmed Words:", stemmed_words)

# ✅ Step 5: Test the function
test_sentence = "NLP techniques are used in virtual assistants like Alexa and Siri."
nlp_preprocess(test_sentence)
```
